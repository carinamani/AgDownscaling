```python
##### Add ISO3 codes to country shapefiles  

import os
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
##### Load data

# Get the current working directory
cd = os.path.dirname(os.getcwd())

countries_simple = gpd.read_file("/Users/carinamanitius/Documents/Data/Admin_Boundaries/simple_for_map/World_Countries_Generalized.shp")

country_codes = pd.read_csv(f"{cd}/Data/Correspondence_tables/country_names.csv", encoding="cp1252")

save_path = f"{cd}/Data/Clean/Geographies/country_geo.shp"
```

```python
#### Merge and resave 

# add in Namibia manually 
country_codes.loc[country_codes['Country_Name'] == 'Namibia', 'ISO2'] = 'NA'

countries_simple = countries_simple.merge(country_codes, left_on='ISO', right_on='ISO2', how='left')

countries_simple.to_file(save_path)
```
