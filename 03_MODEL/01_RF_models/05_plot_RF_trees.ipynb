```python
##### Evaluate RF models 
# calculates model performance metrics 

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.tree import plot_tree
import joblib
import glob
import matplotlib.pyplot as plt
```

```python
##### SET-UP

# Get the current working directory
cd = Path.cwd().parent.parent 

# import models
capital_rf = joblib.load(f'{cd}/Results/RF_models_final/capital_rf_final/model.joblib')
labor_rf = joblib.load(f'{cd}/Results/RF_models_final/labor_rf_final/model.joblib')

# set path to figure directory
fd = "/Users/carinamanitius/Library/CloudStorage/OneDrive-UniversityofVermont/Documents/OneDrive/Dissertation/Chapter 1/Figures/RESULTS/RF_trees"
```

```python
##### VISUALIZE TOP OF TREES

# feature names (adjust if your model was trained on a DataFrame with column names)
capital_features = getattr(capital_rf, "feature_names_in_", None)
labor_features = getattr(labor_rf, "feature_names_in_", None)

# --- Capital RF tree ---
fig, ax = plt.subplots(figsize=(14, 10))
plot_tree(
    capital_rf.estimators_[0],
    max_depth=3,
    feature_names=capital_features,
    filled=True,
    rounded=True,
    fontsize=9,
    ax=ax
)
ax.set_title("Capital RF — representative tree (top 3 levels)")
plt.tight_layout()
plt.savefig(f"{fd}/capital_rf_tree_top.png", dpi=300, bbox_inches="tight")
plt.show()

# --- Labor RF tree ---
fig, ax = plt.subplots(figsize=(14, 10))
plot_tree(
    labor_rf.estimators_[0],
    max_depth=3,
    feature_names=labor_features,
    filled=True,
    rounded=True,
    fontsize=9,
    ax=ax
)
ax.set_title("Labor RF — representative tree (top 3 levels)")
plt.tight_layout()
plt.savefig(f"{fd}/labor_rf_tree_top.png", dpi=300, bbox_inches="tight")
plt.show()
```
